### Question 4: Standardize Missing or Invalid Strings (2 Marks)

For all columns with an `object` (string) data type in `df1`, perform the following replacements to standardize missing data:
- Replace the specific character `'–'` with the string `'Unknown'`.
- Replace the string `'nan'` with `'Unknown'`.
- Replace any entirely empty strings or strings consisting only of whitespace with `'Unknown'`.

**Expected Output:** The code cell should execute without errors, standardizing the text data.

```python
# Answer Q4
import numpy as np

# Get all object columns
object_columns = df1.select_dtypes(include='object').columns

for col in object_columns:
    # Replace '–' with 'Unknown'
    # Ensure conversion to string type first to handle mixed types gracefully
    df1[col] = df1[col].astype(str).str.replace('–', 'Unknown', regex=False)

    # Replace 'nan' string with 'Unknown'
    df1[col] = df1[col].str.replace('nan', 'Unknown', regex=False)

    # Replace empty strings or strings consisting only of whitespace with 'Unknown'
    # The regex r'^\s*$' matches strings that are either empty or contain only whitespace characters.
    df1[col] = df1[col].replace(r'^\s*$', 'Unknown', regex=True)

    # Finally, fill any remaining actual numpy.nan values that might exist with 'Unknown'
    # (though previous steps should convert them to string 'nan' then 'Unknown')
    df1[col] = df1[col].fillna('Unknown')



```

### Question 7: Handle Low-Frequency Categories (2 Marks)

Some categorical columns have too many rare values. For both the `transmission` and `ext_col` columns, group categories that appear in **less than 1%** of the total dataset into a new category called `'others'`.
Additionally, drop the `clean_title` column entirely as it provides little variance.

**Expected Output:** The code cell should execute without errors.

```python
 # Answer Q7

# Define a function to group low-frequency categories
def group_low_frequency_categories(df, column, threshold=0.01):
    total_rows = len(df)
    # Calculate value counts and their proportion
    value_counts = df[column].value_counts()
    proportions = value_counts / total_rows

    # Identify categories to be grouped as 'others'
    low_frequency_categories = proportions[proportions < threshold].index

    # Replace low-frequency categories with 'others'
    df[column] = df[column].replace(low_frequency_categories, 'others')
    return df

# Apply the function to 'transmission' column
df1 = group_low_frequency_categories(df1, 'transmission', threshold=0.01)

# Apply the function to 'ext_col' column
df1 = group_low_frequency_categories(df1, 'ext_col', threshold=0.01)

# Drop the 'clean_title' column, if it exists
if 'clean_title' in df1.columns:
    df1 = df1.drop('clean_title', axis=1)
    print("'clean_title' column dropped.")


print("DataFrame after handling low-frequency categories and potentially dropping 'clean_title':")
# Display unique values and their counts for 'transmission' and 'ext_col' to verify (optional)
print("\nUnique values and counts for 'transmission' after grouping:")
display(df1['transmission'].value_counts())
print("\nUnique values and counts for 'ext_col' after grouping:")
display(df1['ext_col'].value_counts())

# Display info to confirm 'clean_title' is dropped and types are still correct (optional)
print("\nDataFrame info after changes:")
df1.info()
```

### Question 8: Remove Outliers and Apply Label Encoding (2 Marks)

Remove price outliers by filtering `df1` to keep only rows where the `price` is **below the 99th percentile**.
Next, import `LabelEncoder` from `sklearn.preprocessing` and apply it to convert all remaining categorical (`object`) columns into numerical values.

**Expected Output:** The code cell should execute without errors.

```python
# Answer Q8
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Re-clean and convert 'milage' and 'price' columns to ensure they are numeric
# This step is added as a safeguard in case previous cleaning steps did not persist or were incomplete.

# For 'milage': Remove ' mi.' and commas, then convert to numeric
# First, ensure it's string to apply string methods reliably
df1.loc[:, 'milage'] = df1['milage'].astype(str).str.replace(' mi.', '', regex=False)
df1.loc[:, 'milage'] = df1['milage'].astype(str).str.replace(',', '', regex=False)
df1.loc[:, 'milage'] = pd.to_numeric(df1['milage'], errors='coerce').astype('Int64') # Using 'Int64' to allow for NaN if coercion fails

# For 'price': Remove currency symbols ($) and commas, then convert to numeric
# First, ensure it's string to apply string methods reliably
df1.loc[:, 'price'] = df1['price'].astype(str).str.replace('$', '', regex=False)
df1.loc[:, 'price'] = df1['price'].astype(str).str.replace(',', '', regex=False)
df1.loc[:, 'price'] = pd.to_numeric(df1['price'], errors='coerce') # Converts to float64 by default if there are non-integer values or NaNs

# 1. Remove price outliers by filtering df1 to keep only rows where the price is below the 99th percentile
# Calculate the 99th percentile of the 'price' column
price_99th_percentile = df1['price'].quantile(0.99)

# Filter the DataFrame to keep only rows where price is below the 99th percentile
df1 = df1[df1['price'] < price_99th_percentile]

print(f"DataFrame shape after removing price outliers: {df1.shape}")

# 2. Apply LabelEncoder to convert all remaining categorical (object) columns into numerical values
# Identify remaining object columns
object_columns_to_encode = df1.select_dtypes(include='object').columns.tolist()

# Explicitly remove 'price' from the list of columns to be label encoded
# as 'price' is intended to be a numeric target variable, not a categorical feature.
if 'price' in object_columns_to_encode:
    print("Removing 'price' from the list of columns to be label encoded as it should be numeric.")
    object_columns_to_encode.remove('price')

# Initialize LabelEncoder
le = LabelEncoder()

for col in object_columns_to_encode:
    # Apply Label Encoding to each object column
    df1.loc[:, col] = le.fit_transform(df1[col])
    print(f"Column '{col}' label encoded.")

print("\nDataFrame info after outlier removal and label encoding:")
df1.info()
print("\nFirst 5 rows after outlier removal and label encoding:")
display(df1.head())
```

### Question 10: Model Training and Evaluation (4 Marks)

Import `RandomForestRegressor` from `sklearn.ensemble`, and `r2_score`, `mean_squared_error`, `mean_absolute_error` from `sklearn.metrics`.
Train the Random Forest model (with `n_estimators=100`, `random_state=42`) on the training data. Predict the prices for the test data and evaluate the model by printing the R² Score, MSE, and MAE.

**Expected Output:** Three numeric values representing the R² score, Mean Squared Error, and Mean Absolute Error of the model.

```python
# Answer Q11
from sklearn.datasets import load_digits

# Load the MNIST dataset
digits = load_digits()

# Separate features (X) and target labels (y)
X = digits.data
y = digits.target

# Print the shape of the features and the target arrays
print(f"Shape of features (X): {X.shape}")
print(f"Shape of target labels (y): {y.shape}")
```

### Question 12: K-Means Clustering (5 Marks)

Import `KMeans` from `sklearn.cluster`. Initialize and fit the K-Means clustering model on the MNIST features (`X`).
Since we know there are 10 digits (0-9), set the number of clusters to 10. Assign the cluster labels to a variable `kmeans_labels`.

**Expected Output:** A successfully fitted K-Means model and the cluster labels array.

```python
# Answer Q12
from sklearn.cluster import KMeans

# Initialize KMeans with 10 clusters (for digits 0-9)
# Set random_state for reproducibility
kmeans = KMeans(n_clusters=10, random_state=42, n_init='auto') # n_init='auto' is recommended for newer scikit-learn versions

# Fit the KMeans model to the features (X)
kmeans.fit(X)

# Assign the cluster labels to kmeans_labels
kmeans_labels = kmeans.labels_

print(f"Shape of kmeans_labels: {kmeans_labels.shape}")
print("First 10 cluster labels:")
display(kmeans_labels[:10])
```

### Question 13: F1 Score Evaluation for Clustering (5 Marks)

Evaluate the clustering performance using the F1 score. Since K-Means assigns arbitrary cluster labels, you will first need to map each cluster label to the most frequent true label in that cluster.
After mapping the labels, calculate and print the macro-averaged F1 score using `sklearn.metrics.f1_score`.

**Expected Output:** The calculated F1 score of the clustering.

```python
# Answer Q13
from sklearn.metrics import f1_score
import numpy as np
from scipy.stats import mode

# Create a mapping from cluster label to true label
# Initialize an empty dictionary to store the mapping
labels_map = {}

# Iterate through each unique cluster label found by KMeans
for i in np.unique(kmeans_labels):
    # Find the true labels (y) for all data points that were assigned to cluster i
    # Then, find the most frequent true label among these points
    # mode()[0][0] returns the most frequent element
    most_frequent_true_label = mode(y[kmeans_labels == i], keepdims=False)[0]
    # Store this mapping: cluster_label -> most_frequent_true_label
    labels_map[i] = most_frequent_true_label

# Apply the mapping to the kmeans_labels to get the predicted labels that align with true labels
predicted_labels = np.array([labels_map[label] for label in kmeans_labels])

# Calculate the macro-averaged F1 score
f1 = f1_score(y, predicted_labels, average='macro')

# Print the F1 score
print(f"Macro-averaged F1 Score for K-Means Clustering: {f1:.4f}")
```
